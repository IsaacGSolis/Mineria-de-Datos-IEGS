```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering
```

```python
X = np.array([
    [20, 100],
    [22, 120],
    [23, 140],
    [45, 700],
    [46, 750],
    [48, 720],
    [60, 1500],
    [62, 1600]
])
nombres = [f"C{i+1}" for i in range(8)]
```

```python
plt.scatter(X[:, 0], X[:, 1])
for nombre, (x, y) in zip(nombres, X):
    plt.annotate(nombre, (x, y))
plt.title("Clientes: Edad vs Gasto Mensual")
plt.xlabel("Edad")
plt.ylabel("Gasto Mensual ($)")
plt.grid(True)
plt.show()
```

```python
Z = linkage(X, method='ward', metric='euclidean')
dendrogram(Z,labels=nombres)
plt.title("Dendrograma — Clientes")
plt.xlabel("Clientes")
plt.ylabel("Distancia")
plt.grid(True)
plt.show()
```

```python
modelo2 = AgglomerativeClustering(n_clusters=2, linkage='ward')
etiquetas2 = modelo2.fit_predict(X)
plt.scatter(X[:, 0], X[:, 1], c=etiquetas2)
for nombre, (x, y) in zip(nombres, X):
    plt.annotate(nombre, (x, y))
plt.title("Resultado de 2 Clusters")
plt.xlabel("Edad")
plt.ylabel("Gasto Mensual ($)")
plt.grid(True)
plt.show()
```

```python
modelo3 = AgglomerativeClustering(n_clusters=3, linkage='ward')
etiquetas3 = modelo3.fit_predict(X)
plt.scatter(X[:, 0], X[:, 1], c=etiquetas3)
for nombre, (x, y) in zip(nombres, X):
    plt.annotate(nombre, (x, y))
plt.title("Resultado de 3 Clusters")
plt.xlabel("Edad")
plt.ylabel("Gasto Mensual ($)")
plt.grid(True)
plt.show()
```

**¿Qué clientes parecen más similares?**

C2 y C3 comparten el grupo de edades jovenes con poco gasto, C4 y C6 comparten el grupo de edades adultas con gastos medios y C7 y C8 comparten el grupo de dades mayores con gastos mayores.

**¿Qué grupos naturales identifica el algoritmo?**

3 grupos, jovenes, adultos y mayores.

**¿Qué representa cada cluster desde el punto de vista comercial?**

Clientes económicos o jovenes.

Clientes estables con consumo moderado.

Clientes premium o de alto poder adquisitivo.

**¿Cuál sería una estrategia de marketing para cada grupo?**

Jovenes: Descuentos y promociones.

Adultos: Programas de fidelidad.

Mayores: Atencion personalizada.

**¿Por qué el algoritmo agrupa ciertos clientes?**

Por las caracteristicas similares que presentan (distancia) como edad y gastos.
