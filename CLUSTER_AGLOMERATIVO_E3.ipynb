```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering
```

```python
X = np.array([
    [95, 98],
    [92, 94],
    [88, 90],
    [65, 70],
    [60, 68],
    [58, 62],
    [30, 40],
    [35, 38]
])
nombres = ["A", "B", "C", "D", "E", "F", "G", "H"]
```

```python
Z = linkage(X, method='ward', metric='euclidean')
print(Z)
```

```python
dendrogram(Z, labels=nombres)
plt.title("Dendrograma — Rendimiento Académico")
plt.xlabel("Alumnos")
plt.ylabel("Distancia")
plt.grid(True)
plt.show()
```

```python
modelo2 = AgglomerativeClustering(n_clusters=2, linkage='ward')
etiquetas2 = modelo2.fit_predict(X)
plt.scatter(X[:, 0], X[:, 1], c=etiquetas2)
for nombre, (x, y) in zip(nombres, X):
    plt.annotate(nombre, (x, y))
plt.title("Resultado de 2 Clusters")
plt.xlabel("Promedio")
plt.ylabel("Asistencia (%)")
plt.grid(True)
plt.show()
```

```python
modelo3 = AgglomerativeClustering(n_clusters=3, linkage='ward')
etiquetas3 = modelo3.fit_predict(X)
plt.scatter(X[:, 0], X[:, 1], c=etiquetas3)
for nombre, (x, y) in zip(nombres, X):
    plt.annotate(nombre, (x, y))
plt.title("Resultado de 3 Clusters")
plt.xlabel("Promedio")
plt.ylabel("Asistencia (%)")
plt.grid(True)
plt.show()
```

**¿Qué alumnos se parecen más?**

A y B, D y E, y por ultimo, G y H.

**¿Qué perfiles de estudiantes aparecen?**

Los 3 clusters que aparecen:

Alto rendimiento.

Rendimiento medio.

Bajo rendimiento.

**¿Qué decisión institucional podría tomarse con esta información?**

Ofrecer tutorias a estudiantes de bajo desempeño, ofrecer becas a estudiantes de alto desempeño, etc.

**¿Qué ocurre cuando el número de clusters cambia?**

Los grupos aumentan tamaño o disminuyen tamaño dependiendo si se aumenta o disminuye el numero de clusters.

**¿Qué grupo requeriría más intervención académica?**

El cluster 2 conformado por G y H que persentan bajo rendimiento academico por tener bajo promedio y bajo porcentaje de asistencia.
