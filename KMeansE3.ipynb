```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
#1 Crear dataframe
data={
    'ventas':[2, 3, 2.5, 15, 18, 20, 5, 6, 7],
    'ingresos':[3, 4, 3.5, 9, 10, 10, 5, 6, 7]
}
df=pd.DataFrame(data)
df
```

```python
#2 Graficar los datos
plt.scatter(df['ventas'], df['ingresos'])
plt.xlabel('Ventas')
plt.ylabel('Ingresos')
plt.show()
```

```python
#3 aplicar KMeans con k=3
kmeans=KMeans(n_clusters=3, random_state=42)
#4 Asignar clusters
df['cluster']=kmeans.fit_predict(df)
df
```

```python
#5 Obtener centroides
centroides=kmeans.cluster_centers_
centroides
```

```python
#6 Visualizar clusters
colores=['red', 'green', 'blue']
for i in range(3):
    cluster=df[df['cluster']==i]
    plt.scatter(cluster['ventas'], cluster['ingresos'], color=colores[i], label=f'Cluster {i}')
    plt.scatter(centroides[i, 0], centroides[i, 1], c=colores[i], marker='X',s=100,label=f'Centroides {i}')
plt.xlabel('Ventas')
plt.ylabel('Ingresos')
plt.title('Clusters')
plt.legend()
plt.show()
```
